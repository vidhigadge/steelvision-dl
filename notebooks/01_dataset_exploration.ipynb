```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

DATA_DIR = Path("../data/raw/NEU-CLS")
```

```python
class_folders = sorted(
    folder for folder in DATA_DIR.iterdir()
    if folder.is_dir()
)

class_names = [folder.name for folder in class_folders]

class_names
```

```python
fig, axes = plt.subplots(2, 3, figsize=(10, 7))

for ax, folder in zip(axes.flat, class_folders):
    image_path = list(folder.glob("*.*"))[0]
    image = Image.open(image_path)

    ax.imshow(image, cmap="gray")
    ax.set_title(folder.name)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
